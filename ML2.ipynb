```python
from sklearn.preprocessing import LabelEncoder

# create LabelEncoder object
encoder = LabelEncoder()

# Encode categorical column (e.g., 'Flight Status')
df['Flight Status Code'] = encoder.fit_transform(df['Flight Status'])

# Display result
print(df[['Flight Status', 'Flight Status Code']].head())
```

```python
from sklearn.preprocessing import MinMaxScaler

# Select only actual numeric columns for scaling
# 'Age' is numeric, and we can include 'Flight Status Code' which we encoded in the previous step
features = ['Age', 'Flight Status Code']
X = df[features]

# Initialize and apply MinMaxScaler
scaler = MinMaxScaler()
rescaled_data = scaler.fit_transform(X)

# Create a new dataframe with the rescaled data
df_rescaled = pd.DataFrame(rescaled_data, columns=X.columns)

# Display result
print(df_rescaled.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Select multiple numeric features so PCA has enough dimensions
# We will use 'Age' and the encoded 'Flight Status Code'
X = df[['Age', 'Flight Status Code']]

# Handle missing values
X = X.fillna(X.mean())

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Perform PCA (n_components matches the number of features)
pca = PCA(n_components=2)
principle_components = pca.fit_transform(X_scaled)

# Plot
plt.scatter(principle_components[:, 0], principle_components[:, 1], alpha=0.5)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Airlines data")
plt.show()
```
